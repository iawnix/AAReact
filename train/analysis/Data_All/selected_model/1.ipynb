```python
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd
```

```python
x = np.random.uniform(0, 20, 10)
y = 0.1*x + 0.5 + np.random.normal(0, 5, 10)
```

```python
k1, b1 = np.polyfit(x, y, deg=1)
print("k = {}, b = {}".format(k1, b1))
k2, b2 = np.polyfit(y, x, deg=1)
print("k = {}, b = {}".format(k2, b2))
```

```python
plt.plot(x, y, 'o')
plt.plot(x, k1*x + b1, color='red', label = "y = {}*x + {}".format(k1, b1))
plt.plot(k2*y + b2, y, color='blue', label = "x = {}*y + {}".format(k2, b2))
plt.legend()
```

```python
r = np.corrcoef(x, y)[0, 1]
print("r = {}".format(r))
r = np.corrcoef(y, x)[0, 1]
print("r = {}".format(r))
```
